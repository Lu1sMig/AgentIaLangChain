## 1.3 Inicializando el modelo LLM

```python
#1.4 Iniciando el modelo LLM
# Instanciación del modelo de lenguaje dentro de la estructura de LangChain
# gemini-3.6-flash usa sampling fijo: no acepta temperature ni top_p
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
```

```python
from IPython.display import Markdown, display
def extraer_texto(respuesta):
    """Aplana el content de un AIMessage.

    Los modelos con razonamiento devuelven una lista de bloques
    ({'type': 'text', ...}, firmas de pensamiento, etc.).
    Los modelos clásicos devuelven un string plano.
    """
    if isinstance(respuesta.content, str):
        return respuesta.content

    return "".join(
        bloque["text"]
        for bloque in respuesta.content
        if isinstance(bloque, dict) and bloque.get("type") == "text"
    )
def mostrar(texto):
    """Renderiza Markdown en la salida de Colab."""
    display(Markdown(texto))
```

## 1.5 Diseño del PrompTemplate

```python
#1.5. Diseño del PromptTemplate e Invocación para TEXTO
#Se construye una plantilla reutilizable mediante PromptTemplate, definiendo explícitamente las variables de entrada dinámicas (input_variables).
from langchain_core.prompts import PromptTemplate
prompt_template = (
    "Eres un experto en {tema} y debes entregar una explicación clara "
    "y actualizada sobre los impactos de la Inteligencia Artificial (IA) "
    "en esta área."
)
modelo_de_prompt = PromptTemplate(
    input_variables=["tema"],
    template=prompt_template,
)
```

## 1.6 Ejecución de la cadena

```python
#1.6 Ejecución de la Cadena / Agente:
#Se invoca la ejecución pasando un diccionario con los valores requeridos por el template (en este caso, el tema "educación"):
# Creación de la cadena combinando la plantilla y el LLM
cadena = modelo_de_prompt | llm
respuesta = cadena.invoke({"tema": "educación"})
mostrar(extraer_texto(respuesta))
```

## 1.7 Mejora de la vexto plano para ser almacenado


```python
from langchain_core.output_parsers import StrOutputParser
# El parser aplana los bloques automáticamente: ya no hace falta extraer_texto
cadena_texto = modelo_de_prompt | llm | StrOutputParser()

mostrar(cadena_texto.invoke({"tema": "educación"}))
```

## 1.8 *Ingresandole* otros temas al promt

```python
for tema in ["educación", "salud", "logística"]:
    mostrar(f"## Tema: {tema}")
    mostrar(cadena_texto.invoke({"tema": tema}))
```

# 2 Conectando el modelo con la web

Esta sección de codigo conecta al LLm con la busqueda en la WEB mediante la API de tavily.

```python
#2.1 integrando API tavaly
import os
from google.colab import userdata
from langchain_community.tools.tavily_search import TavilySearchResults
from langchain_core.tools import tool

# Configuración de la API Key en el entorno
os.environ["TAVILY_API_KEY"] = userdata.get("Tavily")
```

```python
# 2.2 Definición y Decoración de una Herramienta Customizada (@tool)
@tool
def busca_web(query: str) -> list:
    """Realiza una búsqueda en la web sobre un tema específico"""
    tavily_search = TavilySearchResults(
        max_results=2,        # Limita la cantidad de resultados por consulta
        search_depth="advanced", # Profundidad de búsqueda avanzada
        max_tokens=1000       # Controla la extensión máxima devuelta por resultado
    )
    resultado_busca = tavily_search.invoke(query)
    return resultado_busca
```

```python
# Prueba independiente de la herramienta sin pasar por el LLM
busca_web.invoke("impacto de la inteligencia artificial en la educación")
```

```python
# 2.3  Agrupamiento de herramientas disponibles
tools = [busca_web]

# 2. Vinculación del LLM con el conjunto de herramientas
llm_con_herramienta = llm.bind(tools=tools)
```

## 2.3 Nueva sección

```python
#2.4 Construcción del Flujo con LCEL (langchain_core)
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Configuración de la plantilla de prompt
modelo_de_prompt = PromptTemplate(
    template="Usando solo las tools disponibles, dime cuáles son los impactos de la inteligencia artificial en el área de {tema}.",
    input_variables=["tema"],
)

# Construcción de la cadena LCEL
cadena = modelo_de_prompt | llm_con_herramienta | StrOutputParser()

# Invocación de la cadena
respuesta = cadena.invoke({"tema": "educación"})
```
